### Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(sales_df.drop("Purchase", axis=1), sales_df["Purchase"], test_size=0.2, random_state=42)

X_train.shape, X_test.shape
```

```python
X_train.info()
```

### Let's Make Pipeline

#### 1. Drop User_ID Column

```python
from sklearn.preprocessing import FunctionTransformer

trf1 = FunctionTransformer(
    lambda x: x.drop(columns=["User_ID", "Product_ID"])
)

X_train_temp = trf1.fit_transform(X_train)
X_train_temp.sample(1)
```

#### 2. Transform Columns: Replace 4+ by 4 in Stay_In_Current_City_Years

```python
X_train["Stay_In_Current_City_Years"].unique()
```

```python
# X_train["Stay_In_Current_City_Years"] = X_train["Stay_In_Current_City_Years"].replace("4+", 4)
# X_test["Stay_In_Current_City_Years"] = X_test["Stay_In_Current_City_Years"].replace("4+", 4)
# X_train["Stay_In_Current_City_Years"].unique()

trf2 = FunctionTransformer(
    lambda x: x.assign(Stay_In_Current_City_Years=
        x["Stay_In_Current_City_Years"].replace("4+", 4),
                      Product_Hierarchy=
        x["Product_Category_1"].astype(str) + "_" +
        x["Product_Category_2"].fillna(0).astype(int).astype(str) + "_" +
        x["Product_Category_3"].fillna(0).astype(int).astype(str)
                      )
)

X_train_temp = trf2.fit_transform(X_train)
X_train_temp.sample(1)
```

#### 3. Encoding: Now will apply, One Hot Encoding on Gender and City_Category and Ordinal Encoding on Age column And Handling Missing Values

```python
X_train["Age"].unique()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer

# Ordinal Encoding and One Hot Encoding
trf3 = ColumnTransformer(transformers=[
    ('trf_4_1', OrdinalEncoder(categories = [['0-17', '18-25', '26-35', '36-45', '46-50', '51-55', '55+']]), ['Age']),
    ('trf_4_3', SimpleImputer(strategy="constant", fill_value=0), ['Product_Category_2', 'Product_Category_3']),
    ('trf_4_2', OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore"), ['Gender', 'City_Category', 'Product_Category_1',
                                                                   'Product_Category_2', 'Product_Category_3', 'Product_Hierarchy'])
], remainder="passthrough")

trf3.set_output(transform="pandas")
```

#### 4. Feature Scaling: We do feature scaling

```python
from sklearn.preprocessing import StandardScaler

trf4 = StandardScaler()
```

#### 5. Train: Apply Linear Regression Model

```python
from sklearn.linear_model import LinearRegression

trf5 = LinearRegression()
```

#### 6. Make Pipeline and Model Fit

```python
from sklearn.pipeline import Pipeline

pipe = Pipeline([
    ("trf1", trf1),
    ("trf2", trf2),
    ("trf3", trf3),
    ("trf4", trf4),
    ("trf5", trf5)
])
```

```python
# train the pipeline

pipe.fit(X_train, Y_train)
```

#### 7. Testing & Evaluation: Apply test and evaluate based on different matrices 

```python
Y_pred = pipe.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("MAE:", mean_absolute_error(Y_test, Y_pred))
print("RMSE:", np.sqrt(mean_squared_error(Y_test, Y_pred)))
print("R2:", r2_score(Y_test, Y_pred))
```

```python
train_pred = pipe.predict(X_train)

print("Train R2:", r2_score(Y_train, train_pred))
print("Test R2:", r2_score(Y_test, Y_pred))
```

```python
plt.scatter(Y_test, Y_pred, s=5)

plt.plot(
    [Y_test.min(), Y_test.max()],
    [Y_test.min(), Y_test.max()]
)

plt.xlabel('Actual Purchase')
plt.ylabel('Predicted Purchase')
plt.show()
```

#### 8. Predict: Predict the Purchase Column Value

```python
new_customer_df = pd.read_csv("./dataset/test.csv")

new_customer_df.info()
```

```python
predicted_purchases = pipe.predict(new_customer_df)
```

```python
new_customer_df["Purchase"] = predicted_purchases.round().astype(int)

new_customer_df.sample(5)
```

```python
new_customer_df.to_csv("./output/predicted_purchase.csv", index=False)
```
